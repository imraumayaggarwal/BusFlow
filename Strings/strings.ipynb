#### 2. ACCESSING STRINGS

- There are 2 concepts:<br><span style="color:orange;">1. Indexing</span>

```python
s = 'hello'
# the position of the characters of this string are as follows
# h e l l o
# 0 1 2 3 4
print(s[0])
print(s[2])
print(s[4])
print(s[6]) # -> out of bounds.
```

```python
# similarly we also have negative indexing
#  h  e  l  l  o
# -5 -4 -3 -2 -1
print(s[-1])
print(s[-4])
print(s[0]) # -> we can also use this
print(s[1])

# useful for the case when we need to access the last elements
```

<span style="color:orange;">2. Slicing</span>

```python
# [START POS : END POS]
c = 'hello world'
# if we just give in start and pos
print(c[0:4])
```

```python
# here we just got hell, but we mentioned upto 4 pos
# h e l l o
# 0 1 2 3 4
# hence we increment the end by 1 to get entire substring
print(c[0:5])
```

```python
# FROM START pos TO END
print(c[2:])
```

```python
# FROM START TO END POS
print(c[:5])
```

```python
# NO START POS AND END POS
print(c[:])
```

```python
# [START : END : STEP ]
print(c[0:8:2])
```

```python
print(c[::2]) #-> will give char at steps of 2
```

```python
# CAN NEGATIVE STEP WORK WITH POSITIVE START AND END?
print(c[0:5:-1])
```

```python
# no it doesn't work
# but it works with negative start and end surely
print(c[-6::2]) # -> returns with the steps up 2
```

```python
# FAMOUS TECHNIQUE TO REVERSE THE STRING
print(c[::-1])
```

```python
# how to reverse a specific part?
print(c[-1:-5:-1])
```

### Difference: isdecimal vs isdigit vs isnumeric
- `isdecimal()` -> only base-10 decimal characters (0-9 and other decimal-script digits).
- `isdigit()` -> decimal digits plus digit-like characters such as superscripts (for example 2 superscript).
- `isnumeric()` -> widest check: digits plus numeric characters like vulgar fractions and Roman numerals.

```python
samples = ["123", "12.5", "²", "Ⅳ", "½"]
for x in samples:
    print(x, "->", x.isdecimal(), x.isdigit(), x.isnumeric())
```

### Strip function

- it is used to remove spaces before and after and it is dangerous

```python
c = "    hello    "
c.strip()
```

```python
# removes characters from the left side (the start) of a string.
name = "   Hello"
print(name.lstrip())
```

```python
# removes characters from the right side (the end) of a string.
name = "Hello   "
print(name.rstrip()) 
```
