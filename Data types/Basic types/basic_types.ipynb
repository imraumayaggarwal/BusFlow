# Basic Data Types in Python

**Goal:** Learn Python's fundamental data types: int, float, complex, bool, and str.

- **int**: whole numbers (e.g., `42`)
- **float**: decimal numbers (e.g., `3.14`)
- **complex**: numbers with real/imaginary parts (e.g., `1+2j`)
- **bool**: true/false values (`True`, `False`)
- **str**: text sequences (e.g., `"hello"`)

## ✅ Integer (int)

Whole numbers, positive or negative.

```python
# Integer examples
print(4)
print(-10)
print(type(4))
```

## ✅ Float (float)

Numbers with decimal points or scientific notation.

```python
# Float examples
print(4.5)
print(1.7e45)  # scientific notation
print(type(4.5))
```

## ✅ Complex (complex)

Numbers with real and imaginary parts.

```python
# Complex examples
print(1 + 2j)
print(3.5 - 4j)
print(type(1 + 2j))
```

## ✅ Boolean (bool)

True or False values.

```python
# Boolean examples
print(True)
print(False)
print(type(True))
```

## ✅ String (str)

Sequences of characters, enclosed in quotes.

```python
# String examples
print('hello world!')
print("hello world!")
print("""hello world!""")
print(type('hello'))
```
