## GUESSING GAME

<h5>we will randomly generate a number and ask the user to guess the number until the user gets it right</h5>

```python
randomnumber = random.randint(1,100)
counter = 1
guess = int(input("Enter the number:"))
while(guess!= randomnumber):
    if(guess < randomnumber):
        print("Go higher")
    else:
        print("Go lower")
    guess = int(input("Enter the number:"))
    counter+=1
print(f"YOU FOUND THE NUMBER!\n But you took {counter} attempts")
```
