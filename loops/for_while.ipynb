### while loop

```python
number = int(input("Enter number:"))
i = 1
while(i < 11):
    print(number, "x", i, "=", number*i)
    i+=1
```

### For loop

```python
#loop runs till (end-1)
num = int(input("Enter number:"))
# range(end pos):
for i in range(num):
    print(i)
```

```python
num = int(input("Enter number:"))
#range(start pos, end pos)
for i in range(1, num):
    print(i)
```

```python
num = int(input("Enter number:"))
#range(start pos, end pos)
for i in range(2, num):
    print(i)
```

```python
num = int(input("Enter number:"))
#range(start pos, end pos, increment i by number)
for i in range(1, num, 2):
    print(i)
```

```python
num = int(input("Enter number:"))
#range(start pos, end pos, increment i by number)
for i in range(num, 0, -2):
    print(i)
```
