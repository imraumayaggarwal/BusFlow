## Break

- **break** statment is used when you need to break out of the loop at a condition

```python
for i in range(1, 100):
    if i==5:
        break
    print(i)
```

- **continue** statement is used when we need to skip an iteration at a specific condition

```python
for i in range(1, 7):
    if i==5:
        continue
    print(i)
```

- **pass** is used when you are unsure about what you need to write as a code, so you can later visit and write your code there

```python
# without pass an error is producded
for i in range(1, 10):
    
```

```python
# with pass there is no error
for i in range(1, 10):
    pass
```
