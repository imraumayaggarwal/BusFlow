# Type Conversion in Python

**Goal:** Understand **implicit** vs **explicit** conversion and how to change types with `int()`, `str()`, etc.

- 🔁 *Implicit conversion* happens automatically (e.g., `int + float`).
- 🧠 *Explicit conversion* is when you call `int()`, `str()`, `list()`, etc.

```python
# Check the type of a value
print(type([1, 2, 3, 4, 5]))  # list

x = 5
print(type(x))              # int

print(type('hello'))        # str
```

## ✅ Implicit conversion (automatic)

When you mix types in an expression, Python converts values to a common type.

```python
# int + float -> float
result = 4 + 3.5
print(result, type(result))

# float + complex -> complex
result = 4.5 + 6j
print(result, type(result))

# int + complex -> complex
result = 5 + 6 + 7j
print(result, type(result))
```

## ✅ Explicit conversion (you choose the type)

Use built-in constructors like `int()`, `str()`, `float()`, `complex()`, and `list()`.

```python
# Convert values explicitly
print(int(4))       # already int
print(int(34.3))    # float -> int (drops fractional part)
print(int(223.3))   # float -> int

print(str(5.4))     # float -> str
print(complex(4))   # int -> complex (4 + 0j)
print(list('hello'))
```
