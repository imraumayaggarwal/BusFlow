# The Print Function in Python

**Goal:** Learn how to display output using `print()`, including separators and endings.

- `print()` outputs text/values to the console.
- Can print multiple items, customize separators (`sep`), and endings (`end`).

## ✅ Basic print usage

Print strings, numbers, or mixed types.

```python
# Simple print
print("Hello World!")

# Print multiple items (separated by space by default)
print("hey", "how", "are", "you")

# Print different data types
print("hey", 5.6, 5)
```

## ✅ Custom separators (`sep`)

Change how items are joined (default is space).

```python
# Custom separator
print("hey", "how", "are", "you", sep='-')
print("hey", "how", "are", "you", sep='/')
```

## ✅ Custom endings (`end`)

Change what comes after the print (default is newline `\n`).

```python
# Custom end (no newline)
print("hello", end=' ')
print("world")  # continues on same line
```

💡 Tip: Press Shift+Tab in VS Code to see `print()` documentation.
