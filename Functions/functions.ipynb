# Functions in Python
Functions help us organize code into reusable blocks. They make programs easier to read, test, and maintain.

## What are Functions?

```python
# A function is a reusable block of code that performs a specific task.
# You define it once and call it whenever you need the same task again.
def greet(name):
    print("Hello", name)
    # The function prints a greeting using the value passed into name.
greet("Aarav")
greet("Maya")
```

## Components of a Function

```python
# Components of a function are: def, function name, parameters, body, and return statement.
def add_numbers(a, b):
    """Return the sum of two numbers."""
    # a and b are parameters; they receive values when the function is called.
    total = a + b
    # total is a local variable, created inside the function.
    return total
result = add_numbers(5, 7)
print(result)
```

## Important Concepts

```python
# Functions can have default values, keyword arguments, and variable scope.
def introduce(name, city="Unknown"):
    # city has a default value, so it is optional when calling the function.
    print(name, "is from", city)

introduce("Riya")
introduce("Arjun", city="Delhi")

score = 100
# score is a global variable because it is defined outside the function.
def show_score():
    # This score is local to the function and does not change the global one.
    score = 50
    print("Local score:", score)

show_score()
print("Global score:", score)
```

## Functions as Objects

```python
# In Python, functions are objects, so they can be stored in variables, passed around, and returned.
def greet(name):
    return "Hello " + name

# Store the function in another variable.
welcome = greet
print(welcome("Sam"))

# Pass a function to another function.
def apply_function(func, value):
    # func is a function object passed as an argument.
    return func(value)

print(apply_function(greet, "Anita"))
```

## Nested Functions

```python
# A nested function is a function defined inside another function.
def outer(name):
    # This is the outer function.
    greeting = "Hello"

    def inner():
        # This is the nested function; it can use variables from the outer function.
        print(greeting, name)

    inner()  # Call the inner function from inside the outer function.

outer("Nina")
```
