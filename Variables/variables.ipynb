# Variables in Python (Dynamic Typing + Dynamic Binding)

**Goal:** Learn why Python variables are labels bound to values, and how you can reassign them freely.

- ✅ No type declarations (types belong to values).
- ✅ A variable is a name that points to a value.
- ✅ You can rebind the same name to different values.

---

## ✅ What is a variable in Python?

A variable is a **name** that points to a **value**.

- The name is like a label (`x`, `name`).
- The value is the thing labeled (`10`, `"hello"`, `[1, 2, 3]`).

You can reassign the name to a different value anytime.

```python
# Example: a variable points to a value (the value has a type)

x = 10          # x points to an integer value (type int)
print("x:", x, "(type", type(x), ")")

x = "hello"    # now x points to a string value (type str)
print("x:", x, "(type", type(x), ")")

x = [1, 2, 3]    # now x points to a list value (type list)
print("x:", x, "(type", type(x), ")")
```

## 🔄 Dynamic Typing (type belongs to the value, not the name)

In Python, types belong to values. The variable name is just a label.

- You don't declare types.
- A variable can refer to different types at different times.

```python
# Dynamic typing demonstration

value = 5              # value is an int
print(value, "->", type(value))

value = 5.0            # now value is a float
print(value, "->", type(value))

value = "Python"      # now value is a string
print(value, "->", type(value))
```

## 🧩 Dynamic Binding (name ↔ value link is made at runtime)

Assignment makes a name point to a value. This binding happens when the code runs.

You can reassign the same name to a new value later.

```python
# Dynamic binding demonstration

name = "Alice"  # name is bound to the string "Alice"
print(name, "->", type(name))

name = 42         # name is now bound to the integer 42
print(name, "->", type(name))

# Two names can point to the same value
first = 100
second = first    # second now points to the same value as first
print(first, second)

# Changing what 'first' points to does not change 'second'
first = 200
print(first, second)
```

```python
a = 10
b = 20
print(a)
print(b)

# special syntax for multiple variable declaration
a= 100;b=200
print(a, b)

# The same variables can also be declared in a single line
a, b = 100, 200
print(a, b)

# the variables can also be declared to each other as 
a = b = c = 20
print(a, b, c)
```
