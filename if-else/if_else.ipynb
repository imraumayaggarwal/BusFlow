# If-Else Statements in Python

**Goal:** Learn how to use conditional statements to control the flow of your program.

- ✅ `if` checks a condition and executes code if true
- ✅ `else` executes code if the condition is false  
- ✅ `elif` checks additional conditions
- ✅ Nested if-else for complex logic

---

```python
email = input("Enter your email: ")
password = input("Enter your password: ")
if email == 'hello@gmail.com' and password == '1234':
    print("successful login!")
else:
    print("Invalid!")
```
