# Tuples in Python
A tuple is an ordered, immutable collection used to store multiple values in a single variable.

## How to Create Tuples

```python
# Tuples are created using parentheses or by separating values with commas.
fruits = ("apple", "banana", "mango")
numbers = 1, 2, 3, 4, 5
single_item = (10,)  # comma is required for a one-item tuple
print(fruits)
print(numbers)
print(single_item)
```

## How to Access Tuples

```python
# Tuples can be accessed using indexing and slicing just like lists.
print(fruits[0])
print(fruits[-1])
print(numbers[1:4])
```

## Editing Tuples

```python
# Tuples are immutable, so we cannot change an item directly.
# fruits[0] = "orange"  # This would raise an error.
old_tuple = (1, 2, 3)
new_tuple = old_tuple[:1] + (99,) + old_tuple[2:]
print(old_tuple)
print(new_tuple)
```

## Deleting Operations and Functions

```python
# We cannot delete a single item from a tuple because tuples are immutable.
sample = (10, 20, 30, 20, 40)
print(len(sample))
print(sample.count(20))
print(sample.index(30))
print(min(sample))
print(max(sample))
print(sorted(sample))
print(sorted(sample, reverse=True))
del sample
# print(sample)  # This would raise an error because the tuple variable has been deleted.
```

## Tuple Operations

```python
# Tuples support operations like concatenation, repetition, and membership tests.
t1 = (1, 2, 3)
t2 = (4, 5)
print(t1 + t2)
print(t1 * 2)
print(2 in t1)
```
