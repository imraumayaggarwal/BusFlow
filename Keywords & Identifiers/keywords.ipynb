# Python Keywords & Identifiers

**Goal:** Learn what Python keywords are, why they are reserved, and how to check them.

- Keywords are special reserved words that have meaning in Python syntax.
- You cannot use keywords as variable names.
- Identifiers are names you choose (variables, functions, etc.) and must not match keywords.

## ✅ See the list of all Python keywords

The `keyword` module provides the current list of reserved Python words.

```python
import keyword

print("Python keywords (", len(keyword.kwlist), "):")
print(keyword.kwlist)
```
