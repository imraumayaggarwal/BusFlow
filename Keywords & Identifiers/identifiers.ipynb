# Python Identifiers (Names in Code)

**Goal:** Learn what identifiers are, basic naming rules, and how to choose clear names.

- Identifiers are names you create (variables, functions, etc.).
- They must follow Python naming rules and cannot be keywords.

## ✅ Identifier rules (what you can name things)

A valid identifier:
- Starts with a letter (`a-z`, `A-Z`) or underscore (`_`)
- Is followed by letters, digits (`0-9`), or underscores
- Cannot start with a digit
- Cannot be a Python keyword (like `for`, `if`, `True`)

```python
# Good identifiers
my_variable = 10
user_name = "Alice"
_count = 5

print(my_variable, user_name, _count)

# Bad identifiers (will cause errors if uncommented)
# 1st_value = 1    # starts with a digit
# my-variable = 2  # hyphen not allowed
# for = 3          # 'for' is a keyword
```

## 🧠 Naming tips

- Use meaningful names (e.g., `total_price`, `user_age`).
- Use underscores to separate words (`snake_case`).
- Avoid single letters except in short loops (`i`, `j`).
