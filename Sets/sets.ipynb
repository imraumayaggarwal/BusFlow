# SETS
- Sets do not allow duplicates. 
- Sets have no indexing/slicing. 
- Sets don't allow mutable data types. ( we cannot insert mutable data types like lists, inside set )
- set itself is a mutable datatype ( i.e 2d sets are not possible )

### EMPTY SET

```python
s = set()
print(type(s))
```

```python
# HOMOGENOUS SET
s1 = {1, 2, 3, 4, 5}
# HETEROGENOUS SET
s2 = {"hellow", 4.5, 2}
print(s1)
print(s2)
```

### CANNOT STORE DUPLICATE VALUES

```python
s3 = {1,1,2,2,3,3,4,4}
print(s3)
```

### CONTAINS ONLY IMMUTABLE DATA TYPES

```python
s4 = {[1,2,3], "hello"}
```

```python
s4 = {(1,2,3), "hello"}
print(s4)
```

### INDEXING NOT ALLOWED

- hashing is followed here

```python
s4[0]
```

```python
s4[0:2]
```

- we cannot edit the items in the set at all

```python
# ONE APPROACH COULD BE:
l1 = list(s1)
print(l1)
l1.append(100)
print(l1)
old_id = id(s1)
s1 = set(l1)
print(s1)
new_id = id(s1)
print(old_id)
print(new_id)

# HERE WE GET DIFFERENT MEMORY ADDRESSES FOR THE SET, WHICH PROVES THAT THE SET IS IMMUTABLE
```

### WE CAN ADD ELEMENTS IN THE SET

```python
s1.add(200)
print(s1)
s1.add(7)
print(s1)
```

### DELETE | REMOVE | POP

```python
print(s4)
del s4
print(s4)
```

```python
print(s1)
s1.remove(200)
print(s1)
s1.pop() # pop() -> removes the very first element in the set
print(s1)
s1.pop()
print(s1)
```

```python
# Two example sets
s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}
print("s1:", s1)
print("s2:", s2)

# union -> all unique elements from both sets
print("Union (s1 | s2):", s1.union(s2))

# intersection -> only common elements present in both sets
print("Intersection (s1 & s2):", s1.intersection(s2))

# symmetric_difference -> elements in either set, but not in both
print("Symmetric Difference (s1 ^ s2):", s1.symmetric_difference(s2))

# difference -> elements present in first set but not in second set
print("Difference (s1 - s2):", s1.difference(s2))
print("Difference (s2 - s1):", s2.difference(s1))

# subset -> checks whether all elements of one set are inside another set
a = {1, 2}
b = {1, 2, 3, 4}
print("a is subset of b:", a.issubset(b))

# superset -> checks whether a set contains all elements of another set
print("b is superset of a:", b.issuperset(a))

# isdisjoint -> checks whether two sets have no common elements
x = {10, 20}
y = {30, 40}
print("x and y are disjoint:", x.isdisjoint(y))
```
