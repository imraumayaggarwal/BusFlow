# Boolean Literals in Python

**Goal:** Learn Python's boolean values (`True` and `False`) and how they work in expressions.

- Booleans represent truth values: `True` (1) or `False` (0).
- Used in conditions, comparisons, and logical operations.

## ✅ Boolean literals

- `True`: represents truth (equivalent to 1 in math).
- `False`: represents falsehood (equivalent to 0 in math).
- Case-sensitive: must be capitalized.

```python
# Boolean examples
print(True)       # boolean True
print(False)      # boolean False

# Check types
print(type(True), type(False))

# Booleans in math (True=1, False=0)
print(True + 1)   # 2
print(False * 5)  # 0
```

## ✅ Comparisons produce booleans

Use `==`, `!=`, `<`, `>`, etc., to get boolean results.

```python
# Comparisons
print(5 == 5)     # True
print(10 > 20)    # False
print("a" != "b") # True

# Logical operations
print(True and False)  # False
print(True or False)   # True
print(not True)        # False
```
