# Special Literals in Python

**Goal:** Learn special values like `None` that represent unique concepts in Python.

- `None`: represents "no value" or absence (like `null` in other languages).
- Used to indicate uninitialized variables or function returns.

## ✅ None literal

- `None` is a singleton object (only one instance exists).
- Represents nothingness or missing data.
- Functions return `None` by default if no explicit return.

```python
# None examples
x = None
print(x)          # None
print(type(x))    # <class 'NoneType'>

# Function without return
def no_return():
    pass

result = no_return()
print(result)     # None

# Check for None
print(x is None)  # True
```
