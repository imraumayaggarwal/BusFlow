# Numeric Literals in Python

**Goal:** Learn how to write numbers in Python code (integers, floats, complex) and special notations.

- Literals are fixed values in code (e.g., `42`, `3.14`, `1+2j`).
- Python supports decimal, binary, octal, hexadecimal, and scientific notation.

## ✅ Integer literals (whole numbers)

- Decimal: `10`, `42`
- Binary: `0b1010` (starts with `0b`)
- Octal: `0o12` (starts with `0o`)
- Hexadecimal: `0xA` (starts with `0x`)

```python
# Integer examples
print(10)        # decimal
print(0b1010)    # binary (10 in decimal)
print(0o12)      # octal (10 in decimal)
print(0xA)       # hex (10 in decimal)

# Check types
print(type(10), type(0b1010))
```

## ✅ Float literals (decimal numbers)

- Standard: `3.14`, `2.0`
- Scientific: `1.23e4` (1.23 × 10^4)
- Underscores for readability: `1_000_000.0`

```python
# Float examples
print(3.14)          # standard float
print(1.23e4)        # scientific notation (12300.0)
print(1_000_000.0)   # underscores for readability

print(type(3.14), type(1.23e4))
```

## ✅ Complex literals (real + imaginary)

- Written as `real + imag j` (e.g., `1 + 2j`)
- `j` represents the imaginary unit (√-1)

```python
# Complex examples
print(1 + 2j)        # complex number
print(3.5 - 4j)      # negative imaginary part
print(5j)            # pure imaginary

print(type(1 + 2j))
```
