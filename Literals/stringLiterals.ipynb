# String Literals in Python

**Goal:** Learn how to write text in Python code with quotes, escapes, and multi-line strings.

- Strings are sequences of characters enclosed in quotes.
- Single (`'`), double (`"`), or triple quotes (`'''` or `"""`) for multi-line.

## ✅ Basic string literals

- `'hello'` or `"hello"` (single or double quotes).
- `'''multi\nline'''` or `"""multi\nline"""` (triple for multi-line).

```python
# String examples
print('hello')          # single quotes
print("world")          # double quotes

# Multi-line
multi = '''This is
a multi-line
string'''
print(multi)

print(type('hello'))
```

## ✅ Escape sequences and raw strings

- `\n` for newline, `\"` for quote inside string.
- Raw strings: `r"no\escapes"` (backslashes treated literally).

```python
# Escapes
print("He said \"hello\"")  # escaped quote
print("Line 1\nLine 2")      # newline

# Raw string
print(r"C:\path\to\file")  # no escape processing

# Formatted strings (f-strings)
name = "Alice"
print(f"Hello, {name}!")   # f-string
```
