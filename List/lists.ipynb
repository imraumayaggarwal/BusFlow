# Lists in Python
A list is an ordered, mutable collection that can store multiple values in a single variable.

## Creating a List

```python
fruits = ["apple", "banana", "mango"]
numbers = [1, 2, 3, 4, 5]
mixed = [1, "hello", 3.5, True]
print(fruits)
print(numbers)
print(mixed)
```

## Accessing a List

```python
print(fruits[0])
print(fruits[2])
print(numbers[-1])
print(fruits[1:3])
```

## Editing a List

```python
numbers = [1, 2, 3, 4, 5]
numbers[1] = 20
print(numbers)
numbers[2:4] = [30, 40]
print(numbers)
```

## Adding Elements to a List

```python
animals = ["cat", "dog"]
# adds a single item at a time
animals.append("rabbit")
print(animals)
# adds item at desired position
animals.insert(1, "parrot")
print(animals)
# adds multiple items at a time
animals.extend([2, "hey"])
print(animals)
```

## Deleting Elements from a List

```python
items = ["pen", "pencil", "eraser", "eraser", "sharpener"]
del items[1]
print(items)
# removes the specific element mentioned
items.remove("eraser")
print(items)
# removes the item at the end
items.pop()
# clears the entire list
items.clear()
print(items)
```

## Different Operations and Functions

```python
values = [10, 20, 30, 20, 40]
len(values)
print(20 in values)
values.count(20)
values.index(30)
values.sort()
print(values)
values.reverse()
print(values)
```

```python
# 1D list: a single level list of values
list_1d = [1, 2, 3, 4, 5]
print("1D List:", list_1d)

# 2D list: a list of lists
list_2d = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print("2D List:")
for row in list_2d:
    print(row)

# Accessing elements in a 2D list
print("2D element at [1][2]:", list_2d[1][2])
print("2D row at [0]:", list_2d[0])

# 3D list: a list of 2D lists
list_3d = [
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
]
print("3D List:")
for matrix in list_3d:
    print(matrix)

# Accessing elements in a 3D list
print("3D element at [0][1][1]:", list_3d[0][1][1])
print("3D matrix at [1]:", list_3d[1])
print("3D row at [0][0]:", list_3d[0][0])
```

```python
# These functions help summarize and order items in a list.
values = [10, 20, 30, 20, 40]
print("Length:", len(values))
print("Minimum:", min(values))
print("Maximum:", max(values))
# this returns us the copy of the list
print("Sorted:", sorted(values))
# this also returns us the copy of the list
print("Sorted in reverse:", sorted(values, reverse=True))
print(values)
```

```python
# for permenant changes
values.sort(reverse=True)
print(values)
values.sort()
print(values)
```

## List Comprehension Examples

```python
# 1) Squares of numbers from 1 to 10
squares = [n * n for n in range(1, 11)]
print("Squares:", squares)

# 2) Even numbers from 1 to 20
evens = [n for n in range(1, 21) if n % 2 == 0]
print("Even numbers:", evens)

# 3) Convert names to uppercase
names = ["riya", "arjun", "sam", "nina"]
upper_names = [name.upper() for name in names]
print("Uppercase names:", upper_names)

# 4) Length of each word in a list
words = ["python", "list", "comprehension", "easy"]
word_lengths = [len(word) for word in words]
print("Word lengths:", word_lengths)

# 5) Flatten a 2D list into a 1D list
matrix = [[1, 2], [3, 4], [5, 6]]
flattened = [item for row in matrix for item in row]
print("Flattened list:", flattened)

# 6) Conditional expression in list comprehension
labels = ["Even" if n % 2 == 0 else "Odd" for n in range(1, 11)]
print("Odd/Even labels:", labels)
```
